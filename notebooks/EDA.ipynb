```python
df['Temperature Difference'] = (
    df['Process temperature [K]']
    - df['Air temperature [K]']
)
```

```python
df['Angular Speed'] = (
    2 * np.pi * df['Rotational speed [rpm]']
) / 60

df['Mechanical Power'] = (
    df['Torque [Nm]']
    * df['Angular Speed']
)
```

```python
sns.boxplot(
    data=df,
    x='Machine failure',
    y='Mechanical Power'
)
```
